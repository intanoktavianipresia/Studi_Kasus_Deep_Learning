```python
# ============================================================
# STUDI KASUS TOPIK 2
# BRIDGING GLOBAL INFRASTRUCTURE TO ADDRESS LOCAL CHALLENGES
# IN HIGH PERFORMANCE COMPUTING (HPC)
# ============================================================

# ============================================================
# SESI 1 — PENDAHULUAN
# ============================================================

"""
Studi kasus:
Analisis efisiensi pemrosesan data kualitas udara.

Tujuan:
1. Menggunakan dataset kualitas udara nyata.
2. Melakukan preprocessing data.
3. Membandingkan waktu pemrosesan beberapa metode.
4. Mengamati efisiensi operasi vektor NumPy.
5. Menghubungkan kebutuhan komputasi dengan analisis
   permasalahan lingkungan lokal.

Catatan:
Eksperimen ini adalah benchmark komputasi dasar, bukan
simulasi cluster HPC atau sistem komputasi terdistribusi.
"""

# ============================================================
# SESI 2 — IMPORT LIBRARY
# ============================================================

import time
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from io import BytesIO
from IPython.display import display

print("Library berhasil disiapkan.")

# ============================================================
# SESI 3 — PENGAMBILAN DATA ASLI
# ============================================================

# Dataset Air Quality dari UCI Machine Learning Repository.
url = "https://archive.ics.uci.edu/static/public/360/air+quality.zip"

response = requests.get(url, timeout=90)
response.raise_for_status()

import zipfile

with zipfile.ZipFile(BytesIO(response.content)) as arsip:
    nama_file = [
        nama for nama in arsip.namelist()
        if nama.lower().endswith(".csv")
        and "airquality" in nama.lower()
    ]

    if not nama_file:
        raise FileNotFoundError(
            "File CSV Air Quality tidak ditemukan dalam arsip UCI."
        )

    with arsip.open(nama_file[0]) as file:
        df = pd.read_csv(file, sep=";", decimal=",")

print("Sumber: UCI Machine Learning Repository")
print("Ukuran dataset:", df.shape)
display(df.head())

# ============================================================
# SESI 4 — PREPROCESSING DATA (PERBAIKAN)
# ============================================================

# Hapus kolom kosong sepenuhnya
df = df.dropna(axis=1, how="all").copy()

# Hapus kolom tanggal dan waktu dari fitur numerik
df = df.drop(columns=["Date", "Time"], errors="ignore")

# Konversi kolom menjadi numerik
for kolom in df.columns:
    df[kolom] = pd.to_numeric(df[kolom], errors="coerce")

# Nilai -200 adalah penanda missing value pada dataset UCI
df = df.replace(-200, np.nan)

# Gunakan hanya fitur numerik
kolom_numerik = df.select_dtypes(include=np.number).columns.tolist()

# Hapus kolom yang seluruh nilainya kosong
df = df.dropna(axis=1, how="all")
kolom_numerik = df.select_dtypes(include=np.number).columns.tolist()

# Isi nilai kosong menggunakan median
df[kolom_numerik] = df[kolom_numerik].fillna(
    df[kolom_numerik].median()
)

# Hilangkan baris yang masih kosong jika ada
df = df.dropna(subset=kolom_numerik).copy()

X = df[kolom_numerik].to_numpy(dtype=np.float64)

print("Jumlah baris:", len(df))
print("Jumlah fitur:", len(kolom_numerik))
print("Kolom yang digunakan:", kolom_numerik)
print("Total nilai kosong:", int(np.isnan(X).sum()))

display(df.head())


# ============================================================
# SESI 5 — PEMROSESAN PYTHON
# ============================================================

def proses_python(data):
    hasil = []

    for baris in data:
        total = 0.0
        for nilai in baris:
            total += nilai
        hasil.append(total / len(baris))

    return np.array(hasil)


mulai = time.perf_counter()
hasil_python = proses_python(X)
waktu_python = time.perf_counter() - mulai

print(f"Waktu Python: {waktu_python:.6f} detik")


# ============================================================
# SESI 6 — PEMROSESAN NUMPY
# ============================================================

mulai = time.perf_counter()
hasil_numpy = np.mean(X, axis=1)
waktu_numpy = time.perf_counter() - mulai

print(f"Waktu NumPy: {waktu_numpy:.6f} detik")


# ============================================================
# SESI 7 — BENCHMARK DAN EVALUASI
# ============================================================

jumlah_pengujian = 5
waktu_python_berulang = []
waktu_numpy_berulang = []

for _ in range(jumlah_pengujian):
    mulai = time.perf_counter()
    proses_python(X)
    waktu_python_berulang.append(time.perf_counter() - mulai)

    mulai = time.perf_counter()
    np.mean(X, axis=1)
    waktu_numpy_berulang.append(time.perf_counter() - mulai)

rata_python = np.mean(waktu_python_berulang)
rata_numpy = np.mean(waktu_numpy_berulang)

speedup = rata_python / rata_numpy if rata_numpy > 0 else np.nan

hasil_benchmark = pd.DataFrame({
    "Metode": ["Perulangan Python", "Operasi NumPy"],
    "Rata-rata Waktu (detik)": [rata_python, rata_numpy]
})

display(hasil_benchmark)

print(f"Speedup NumPy: {speedup:.2f} kali")

selisih = np.max(np.abs(hasil_python - hasil_numpy))
print(f"Selisih maksimum hasil: {selisih:.10f}")


# ============================================================
# SESI 8 — VISUALISASI PERFORMA
# ============================================================

plt.figure(figsize=(9, 5))
plt.bar(
    hasil_benchmark["Metode"],
    hasil_benchmark["Rata-rata Waktu (detik)"]
)
plt.title("Perbandingan Efisiensi Pemrosesan Data Kualitas Udara")
plt.xlabel("Metode Pemrosesan")
plt.ylabel("Rata-rata Waktu (detik)")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# SESI 9 — ANALISIS DAN VISUALISASI DATA
# ============================================================

print("STATISTIK DESKRIPTIF KUALITAS UDARA")
display(df[kolom_numerik].describe().T)

# Pilih kolom yang benar-benar memiliki data valid
fitur = kolom_numerik[0]
data_histogram = df[fitur].dropna()

if len(data_histogram) > 0:
    plt.figure(figsize=(9, 5))
    plt.hist(data_histogram, bins=30)
    plt.title(f"Distribusi Data: {fitur}")
    plt.xlabel(fitur)
    plt.ylabel("Frekuensi")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
else:
    print("Histogram tidak dapat dibuat karena data fitur kosong.")


# ============================================================
# SESI 10 — KESIMPULAN
# ============================================================

print("""
KESIMPULAN

1. Dataset kualitas udara berasal dari UCI Machine Learning
   Repository.
2. Kolom tanggal dan waktu tidak digunakan sebagai fitur numerik.
3. Missing value ditangani sebelum benchmark dilakukan.
4. Waktu pemrosesan Python dan NumPy dibandingkan berdasarkan
   pengukuran aktual.
5. Benchmark ini menunjukkan efisiensi pemrosesan dasar, tetapi
   belum menguji cluster HPC atau komputasi terdistribusi.
""")
```
